```python
def towers_of_hanoi(n, source, destination, auxiliary):
    
    if n == 1:
        print(f"Move disc 1 from {source} to {destination}")
        return

    towers_of_hanoi(n - 1, source, auxiliary, destination)

    print(f"Move disc {n} from {source} to {destination}")

    towers_of_hanoi(n - 1, auxiliary, destination, source)

number_of_discs = 3
towers_of_hanoi(number_of_discs, source='A', destination='C', auxiliary='B')
```
