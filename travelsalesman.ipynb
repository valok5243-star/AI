```python
from itertools import permutations

def calculate_route_distance(route, graph):
    distance = 0

    for i in range(len(route) - 1):
        distance += graph[route[i]][route[i+1]]

    distance += graph[route[-1]][route[0]]
    return distance

def solve_tsp(graph, start_city):
    cities = list(graph.keys())
    cities.remove(start_city)
    
    best_distance = float('inf')
    best_route = None
    
    for perm in permutations(cities):
        current_route = [start_city] + list(perm)
        current_distance = calculate_route_distance(current_route, graph)
        
        if current_distance < best_distance:
            best_distance = current_distance
            best_route = current_route
            
    return best_route, best_distance

distance_graph = {
    'A': {'A': 0, 'B': 10, 'C': 15, 'D': 20},
    'B': {'A': 10, 'B': 0, 'C': 35, 'D': 25},
    'C': {'A': 15, 'B': 35, 'C': 0, 'D': 30},
    'D': {'A': 20, 'B': 25, 'C': 30, 'D': 0}
}

start = 'A'
route, total_distance = solve_tsp(distance_graph, start)

print(f"Shortest Route: {route} -> {start}")
print(f"Total Distance: {total_distance}")
```
