```python
def solve_monkey_banana():
    
    monkey_pos = "door"      
    box_pos    = "window"   
    monkey_on  = "floor"    
    has_banana = False      
    
    banana_pos = "center"    

    print(f"Initial State -> Monkey: {monkey_pos}, Box: {box_pos}, Position: {monkey_on}")

    if monkey_pos != box_pos:
        monkey_pos = box_pos
        print(f"Action: Walk to Box -> Monkey is now at the {monkey_pos}")

    if monkey_pos == box_pos and box_pos != banana_pos:
        box_pos = banana_pos
        monkey_pos = banana_pos  
        print(f"Action: Push Box to Center -> Box and Monkey are now at the {box_pos}")

    if monkey_pos == box_pos and monkey_on == "floor":
        monkey_on = "box"
        print(f"Action: Climb Box -> Monkey is now on top of the {monkey_on}")

    if monkey_on == "box" and box_pos == banana_pos:
        has_banana = True
        print("Action: Grasp Banana -> Success! Monkey got the banana!")

    return has_banana

solve_monkey_banana()
```
