```python
def print_board(board):
    print(f"\n {board[0]} | {board[1]} | {board[2]} ")
    print("---+---+---")
    print(f" {board[3]} | {board[4]} | {board[5]} ")
    print("---+---+---")
    print(f" {board[6]} | {board[7]} | {board[8]} \n")

def check_winner(b, p):
    # All 8 possible winning line combinations
    win_conditions = [(0,1,2), (3,4,5), (6,7,8), (0,3,6), (1,4,7), (2,5,8), (0,4,8), (2,4,6)]
    return any(b[i] == b[j] == b[k] == p for i, j, k in win_conditions)

def tic_tac_toe():
    # Initialize the board with empty spaces
    board = [" "] * 9
    current_player = "X"
    
    for turn in range(9):
        print_board(board)
        
        # Get valid input from the user
        while True:
            try:
                move = int(input(f"Player {current_player}, choose a spot (0-8): "))
                if 0 <= move <= 8 and board[move] == " ":
                    board[move] = current_player
                    break
                print("Invalid spot or already taken. Try again.")
            except ValueError:
                print("Please enter a valid number between 0 and 8.")
        
        # Check if the current move won the game
        if check_winner(board, current_player):
            print_board(board)
            print(f" Player {current_player} wins!")
            return
            
        # Switch player turns
        current_player = "O" if current_player == "X" else "X"
        
    print_board(board)
    print(" It's a draw!")

tic_tac_toe()
```
