```python
from collections import deque

def solve_water_jug(amt1, amt2, target):
    j1_max, j2_max = 4, 3 
    
    visited = set([(amt1, amt2)])
    queue = deque([(amt1, amt2, [])])
    
    while queue:
        j1, j2, path = queue.popleft()
        if j1 == target or j2 == target:
            return path + [(j1, j2)]
            
        moves = [
            (j1_max, j2, "Fill Jug 1"),
            (j1, j2_max, "Fill Jug 2"),
            (0, j2, "Empty Jug 1"),
            (j1, 0, "Empty Jug 2"),
            # Pour Jug 1 into Jug 2
            (j1 - min(j1, j2_max - j2), j2 + min(j1, j2_max - j2), "Pour Jug 1 -> Jug 2"),
            # Pour Jug 2 into Jug 1
            (j1 + min(j2, j1_max - j1), j2 - min(j2, j1_max - j1), "Pour Jug 2 -> Jug 1")
        ]
        
        for next_j1, next_j2, action in moves:
            if (next_j1, next_j2) not in visited:
                visited.add((next_j1, next_j2))
                queue.append((next_j1, next_j2, path + [(j1, j2, action)]))
                
    return None

solution = solve_water_jug(0, 0, 2)

for step in solution[:-1]:
    print(f"Jugs: ({step[0]}, {step[1]}) -> Action: {step[2]}")
print(f"Final Goal Reached: {solution[-1]}")
```
